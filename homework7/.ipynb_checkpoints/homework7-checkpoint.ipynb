```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Matplotlib

def make_sine_wave(x, A, w):
    return A * np.sin(w * x)

x = np.linspace(0, 2 * np.pi, 1000)

amplitude = np.array([0.5, 1, 1.5, 2, 2.5])
frequency = np.array([1, 2, 3, 4, 5])

y = make_sine_wave(x, amplitude[0], frequency[0])

```
